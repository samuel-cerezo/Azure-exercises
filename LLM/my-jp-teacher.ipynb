```python
import os
from openai import AzureOpenAI
from azure.identity import DefaultAzureCredential, get_bearer_token_provider
```

```python
endpoint = "https://aiazurehub0360805331.openai.azure.com/"
deployment = "gpt-35-turbo"
api_key = "###"
```

```python
token_provider = get_bearer_token_provider(
    DefaultAzureCredential(),
    "https://cognitiveservices.azure.com/.default")
      
 
```

```python

client = AzureOpenAI(
    api_key=api_key,  
    api_version="2024-02-01",
    azure_endpoint = endpoint
)    
```

```python
# Specify the path to your file
file_path = os.path.join("system.txt")

# Open and read the file
with open(file_path, "r", encoding="utf-8") as file:
    prompt_document = file.read()
```

```python
completion = client.chat.completions.create(
    model=deployment,
    messages= [
        {"role": "system", "content" : prompt_document},
        {"role": "user", "content": "Hello"},
        {"role": "assistant", "content": "I don't have time for greetings. If you're here to learn, pay attention.\n- \u65e5\u672c\u8a9e (nihongo) - Japanese language\n- \u5b66\u3076 (manabu) - to learn"}
    ],
    max_tokens=800,
    temperature=0.7,
    top_p=0.95,
    frequency_penalty=0,
    presence_penalty=0,
    stop=None,
    stream=False
)
print(completion.to_json())
```
